# BMERA Simulator

Four-qubit example reporting one Pauli-Z expectation per wire.

```python
import jax
import jax.numpy as jnp

from bmera import build_bmera_meta, forward_batch, forward_single

with jax.default_device(jax.devices("cpu")[0]):
    meta = build_bmera_meta(4, type=2)
    params = jnp.full((meta["depth"], 4, meta["n_per_block"]), 0.1)
    x = jnp.array([0.1, -0.2, 0.3, -0.4])
    single = forward_single(params, x, meta)
    batch = forward_batch(params, jnp.stack([x, -x]), meta)

assert jnp.allclose(single, batch[0])
print("Single sample:", single)
print("Batch shape:", batch.shape)
```

The final gate is CNOT for type 1, CRX for type 2, and CRZ for type 3. In all three types, the lower wire controls the upper wire.

See [validation](BMERA_validation.ipynb) for numerical checks and [research experiments](BMERA_simulator_exec.ipynb) for benchmarks.
